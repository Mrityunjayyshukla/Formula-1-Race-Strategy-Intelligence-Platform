Data Fetching will be performed through following steps.

* Outer loop (Years): Iterates through the years.
* Middle loop (Races): Iterates through each Grand Prix in that year.
* Inner loop (Sessions): Iterates through each session in that particular Grand Prix

```python
import os
import logging

logging.getLogger("fastf1").setLevel(logging.WARNING)
# Enable the FastF1 cache to prevent redundant network request
cache_dir = "f1_ingest_cache"
os.makedirs(cache_dir, exist_ok=True)
f1.Cache.enable_cache(cache_dir)

# Define ingestion parameters
base_output_dir = "f1_parquet_data"
year = 2025
# Automation Loop
print(f"Processing year: {year}")
try:
    # Fetching the event schedule
    schedule = f1.get_event_schedule(year)
except Exception as e:
    print(f"Could not fetch for year: {year}")

# Loop through each round in season schedule
for _,event in schedule.iterrows():
    round_number = event["RoundNumber"]
    race_name = event["EventName"].replace(" ", "_")

    # Skip pre season testing rounds
    if round_number == 0:
        continue
    if round_number < 16:
        continue

    print(f"Fetching {year} Round {round_number}: {race_name} [Race]")

    try:
        # Load session object
        session = f1.get_session(year, round_number, "S")
        session.load()

        df_laps = session.laps
        df_weather = session.weather_data
        df_results = session.results

        save_dir = os.path.join(
        base_output_dir, str(year), f"Round_{round_number}_{race_name}", "R")
        os.makedirs(save_dir, exist_ok=True)

        # Save as optimized parquet file
        if not df_laps.empty:
            df_laps.to_parquet(os.path.join(save_dir, "laps.parquet"))

        if not df_weather.empty:
            df_weather.to_parquet(os.path.join(save_dir, "weather.parquet"))

        if not df_results.empty:
            # Convert columns to string safely if mixed data types cause Parquet serialization warnings
            df_results = df_results.astype(str)
            df_results.to_parquet(os.path.join(save_dir, "results.parquet"))

    except Exception as e:
        print(f"Error processing {2023} Round {round_number} (R): {e}")
        continue

print("\n--- Data Ingestion Complete! ---")
```
